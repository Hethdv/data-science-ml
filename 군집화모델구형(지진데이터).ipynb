```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans


X = np.array([
    [6,3], [11,15],[17,12],[24,10],[20,25],[22,30],[85,70],[60,79],[56,52],[81,91],[80,81]
])

kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
ct = kmeans.cluster_centers_
print(ct)
```

```python
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
ct = kmeans.cluster_centers_
print(ct)
```

```python
print(kmeans.labels_)
plt.scatter(X[:,0],X[:,1],c=kmeans.labels_,cmap='rainbow')
plt.scatter(ct[:,0],ct[:,1],c='red', marker='*', s=150)
```
